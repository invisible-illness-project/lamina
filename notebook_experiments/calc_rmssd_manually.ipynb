# RMSSD 

Example: https://heartratehero.com/RMSSD

```python
def get_square(x: list):
    squares = []
    for i in x:
        squared = i ** 2
        squares.append(squared)

    return squares
```

```python
numbers = [30, -60, 90, -70, 20, 30, -60, 45, -30]
squared = get_square(numbers)
squared
```

```python
def get_sum(x: list):
    return sum(x)
```
