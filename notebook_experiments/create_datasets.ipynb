```python
def check_order(s_df):
    # s is a Series here
    s = s_df.loc[s_df['label'].isin([3, 4]), 'label']

    # Group by runs where consecutive values are equal
    run_id = s.ne(s.shift()).cumsum()

    g = s.groupby(run_id)

    runs = pd.DataFrame({
        'label': g.first(),
        'start_idx': g.apply(lambda x: x.index[0]),
        'end_idx': g.apply(lambda x: x.index[-1]),
        'length': g.size()
    }).reset_index(drop=True)

    print(runs.head(10))

    run_starts = s[s.ne(s.shift())]
    run_order = run_starts.to_numpy()
    run_start_indices = run_starts.index.to_numpy()

    print(run_order[:20])
    print(run_start_indices[:20])
```

```python
labels = new_signal_df['subject'].unique()
for label in labels:
    print(label)
    s_df = DataProcessing.get_specific_subject_df(new_signal_df, label)
    check_order(s_df)
```
